```python
from pathlib import Path
import cv2
from ultralytics import YOLO

# تحميل النموذج
model = YOLO("yolo11s.pt")

# تحديد مسار الصورة
source_image = Path("../images/car5.jpg")

# تنفيذ الاكتشاف
results = model.predict(
    source=str(source_image),
    conf=0.35,
    iou=0.55,
    imgsz=1280,
    classes=[0, 1, 2, 3, 4, 5, 6, 7],
    max_det=100,
    line_width=1
)

# تحديد مجلد المخرجات
output_folder = Path(
    r"C:\Users\Dell\Desktop\AI_Car_Detection\outputs"
)

# إنشاء المجلد تلقائياً
output_folder.mkdir(parents=True, exist_ok=True)

# استخراج اسم الصورة الأصلية
image_name = source_image.stem

# تحديد اسم ملف النتيجة
output_file = output_folder / f"{image_name}_detection.jpg"

# حفظ الصورة بعد الاكتشاف
success = cv2.imwrite(
    str(output_file),
    results[0].plot()
)

# عرض الصورة
results[0].show()

if success:
    print("تم اكتشاف الأجسام وحفظ الصورة بنجاح!")
    print("مكان الحفظ:", output_file)
else:
    print("فشل حفظ الصورة!")
```

```python
from collections import Counter

counts = Counter()

for box in results[0].boxes:
    class_id = int(box.cls[0])
    class_name = results[0].names[class_id]
    counts[class_name] += 1

print("نتائج اكتشاف الأجسام:")

for name, count in counts.items():
    print(f"{name}: {count}")
```
